# 1. select customers names together with each order by the customer

```python

query = '''select c.CustomerName, o.OrderID
from customers as c
join orders as o
using (CustomerID)'''

cur.execute(query)
data = cur.fetchall()

df = pd.DataFrame(data, columns = ['Name', 'order_id'])
df
```

# 3. select customers who did not placed any order yet

# 6. Select customer together with the products that he bought

```python
query = '''select c.CustomerID, c.CustomerName, p.ProductName, od.OrderID
from customers as c
join orders using (CustomerID)
join order_details as od using (OrderID)
join products as p using (ProductID)
order by c.CustomerID asc'''

cur.execute(query)
data = cur.fetchall()

df = pd.DataFrame(data, columns = ['customer_id', 'customer_name', 'product_name', 'order_id'])
df
```

# 7. Select product names together with the name of corresponding category

```python
query = '''select p.ProductID, p.ProductName, c.CategoryName
from products as p
join categories as c
using (CategoryID)
order by p.ProductID asc'''

cur.execute(query)
data = cur.fetchall()

df = pd.DataFrame(data, columns = ['product_id', 'product_name', 'category_name'])
df
```

# 8. Select orders together with the name of the shipping company

```python
query = '''select o.OrderID, o.CustomerID, o.EmployeeID,o.OrderDate, s.ShipperName
from orders as o
join shippers as s 
using (ShipperID)
order by o.OrderID'''

cur.execute(query)
data = cur.fetchall()

df = pd.DataFrame(data, columns = ['order_id', 'customer_id', 'Employee_id', 'order_date', 'shipper_name'])
df
```

# 19. Select orderID together with the total price of  that Order, order the result by total price of order in increasing order

```python
query = '''select od.OrderID, sum(od.Quantity * p.Price) as total_price
from order_details as od
join products as p 
using (ProductID)
group by od.OrderID
order by total_price asc
'''

cur.execute(query)
data = cur.fetchall()

df = pd.DataFrame(data, columns = ['order_id', 'total_price'])
df
```

# 20. Select customer who spend the most money

```python
query = '''select c.CustomerID, c.CustomerName, sum(od.Quantity * p.Price) as total_price
from customers as c
join orders as o using (CustomerID)
join order_details as od using (OrderID)
join products as p using (ProductID)
group by c.CustomerID
order by total_price desc
'''

cur.execute(query)
data = cur.fetchall()

df = pd.DataFrame(data, columns = ['customer_id', 'customer_name','total_price'])
df
```
